### Análise Exploratória de Dados

Nesta etapa, realizamos a análise estatística comparativa entre os grupos A e B, geramos visualizações para identificar padrões e salvamos os resultados em arquivos externos.

```python
import pandas as pd
import matplotlib.pyplot as plt

df_a = pd.read_csv('group-a-1.txt')
df_b = pd.read_csv('group-b-1.txt')

desc_a = df_a.describe()
desc_b = df_b.describe()

desc_a.to_csv('summary_a.csv')
desc_b.to_csv('summary_b.csv')

plt.figure(figsize=(10, 6))
plt.boxplot([df_a['Steps'], df_b['Steps']], tick_labels=['Group A', 'Group B'])
plt.title('Steps Comparison')
plt.savefig('steps_boxplot.png')
plt.close()

plt.figure(figsize=(10, 6))
plt.boxplot([df_a['BMI'], df_b['BMI']], tick_labels=['Group A', 'Group B'])
plt.title('BMI Comparison')
plt.savefig('bmi_boxplot.png')
plt.close()

plt.figure(figsize=(10, 6))
plt.scatter(df_a['Steps'], df_a['BMI'], alpha=0.5, label='Group A')
plt.scatter(df_b['Steps'], df_b['BMI'], alpha=0.5, label='Group B')
plt.xlabel('Steps')
plt.ylabel('BMI')
plt.title('Steps vs BMI')
plt.legend()
plt.savefig('steps_vs_bmi_scatter.png')
plt.close()
```
