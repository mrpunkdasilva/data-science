### Background

You will continue working with the Wisconsin Diagnostic Cancer Dataset, this time implementing Support Vector Machines using scikit-learn.

- Create a simple visualization of the decision boundary using two selected features

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (accuracy_score, precision_score, recall_score,
                             f1_score, classification_report, confusion_matrix,
                             ConfusionMatrixDisplay)

os.makedirs('data', exist_ok=True)
np.random.seed(42)
```

# 1. Load and Explore the Dataset


```python
raw_data = load_breast_cancer(as_frame=True)
df = raw_data.data
df['target'] = raw_data.target
df['diagnosis'] = df['target'].map({0: 'Malignant', 1: 'Benign'})

print(f'Dataset shape: {df.shape}')
print(f'\nTarget distribution:')
print(df['diagnosis'].value_counts())
print(f'\nFeature names ({len(raw_data.feature_names)}):')
print(raw_data.feature_names)
df.head()
```

# 2. Split and Scale the Data


```python
X = df.drop(['target', 'diagnosis'], axis=1)
y = df['target']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print(f'Training set: {X_train_scaled.shape[0]} samples')
print(f'Test set: {X_test_scaled.shape[0]} samples')
```

# 3. SVM with Linear Kernel


```python
svm_linear = SVC(kernel='linear', random_state=42)
svm_linear.fit(X_train_scaled, y_train)

y_pred_linear = svm_linear.predict(X_test_scaled)

acc_linear = accuracy_score(y_test, y_pred_linear)
prec_mal_linear = precision_score(y_test, y_pred_linear, pos_label=0)
rec_mal_linear = recall_score(y_test, y_pred_linear, pos_label=0)
f1_mal_linear = f1_score(y_test, y_pred_linear, pos_label=0)

print(f'Accuracy (Linear): {acc_linear:.4f}')
print(f'Precision (Malignant): {prec_mal_linear:.4f}')
print(f'Recall (Malignant): {rec_mal_linear:.4f}')
print(f'F1-Score (Malignant): {f1_mal_linear:.4f}')

print(f'\nSupport vectors: {svm_linear.n_support_} (total {svm_linear.n_support_.sum()})')
print(f'C: {svm_linear.C}')
print(f'\nClassification Report (Linear):')
print(classification_report(y_test, y_pred_linear, target_names=raw_data.target_names))

print(f'Confusion Matrix (rows = true, cols = predicted):')
print(confusion_matrix(y_test, y_pred_linear))
```

# 4. SVM with RBF Kernel


```python
svm_rbf = SVC(kernel='rbf', random_state=42)
svm_rbf.fit(X_train_scaled, y_train)

y_pred_rbf = svm_rbf.predict(X_test_scaled)

acc_rbf = accuracy_score(y_test, y_pred_rbf)
prec_mal_rbf = precision_score(y_test, y_pred_rbf, pos_label=0)
rec_mal_rbf = recall_score(y_test, y_pred_rbf, pos_label=0)
f1_mal_rbf = f1_score(y_test, y_pred_rbf, pos_label=0)

print(f'Accuracy (RBF): {acc_rbf:.4f}')
print(f'Precision (Malignant): {prec_mal_rbf:.4f}')
print(f'Recall (Malignant): {rec_mal_rbf:.4f}')
print(f'F1-Score (Malignant): {f1_mal_rbf:.4f}')

print(f'\nSupport vectors: {svm_rbf.n_support_} (total {svm_rbf.n_support_.sum()})')
print(f'C: {svm_rbf.C}')
print(f'gamma: {svm_rbf._gamma:.4f} (default = 1 / (n_features * X_scaled.var()))')
print(f'\nClassification Report (RBF):')
print(classification_report(y_test, y_pred_rbf, target_names=raw_data.target_names))

print(f'Confusion Matrix (rows = true, cols = predicted):')
print(confusion_matrix(y_test, y_pred_rbf))
```

# 5. Comparison with Logistic Regression


```python
log_reg = LogisticRegression(max_iter=1000, random_state=42)
log_reg.fit(X_train_scaled, y_train)

y_pred_lr = log_reg.predict(X_test_scaled)

acc_lr = accuracy_score(y_test, y_pred_lr)
prec_mal_lr = precision_score(y_test, y_pred_lr, pos_label=0)
rec_mal_lr = recall_score(y_test, y_pred_lr, pos_label=0)
f1_mal_lr = f1_score(y_test, y_pred_lr, pos_label=0)

print(f'Accuracy (Logistic Regression): {acc_lr:.4f}')
print(f'Precision (Malignant): {prec_mal_lr:.4f}')
print(f'Recall (Malignant): {rec_mal_lr:.4f}')
print(f'F1-Score (Malignant): {f1_mal_lr:.4f}')

print(f'\nClassification Report (Logistic Regression):')
print(classification_report(y_test, y_pred_lr, target_names=raw_data.target_names))

print(f'Confusion Matrix (rows = true, cols = predicted):')
print(confusion_matrix(y_test, y_pred_lr))
```

# 6. Performance Comparison


```python
results = pd.DataFrame({
    'Modelo': ['SVM Linear', 'SVM RBF', 'Regressão Logística'],
    'Acurácia': [acc_linear, acc_rbf, acc_lr],
    'Precisão (Maligno)': [prec_mal_linear, prec_mal_rbf, prec_mal_lr],
    'Recall (Maligno)': [rec_mal_linear, rec_mal_rbf, rec_mal_lr],
    'F1 (Maligno)': [f1_mal_linear, f1_mal_rbf, f1_mal_lr]
})
results = results.sort_values('Acurácia', ascending=False).reset_index(drop=True)

print(results.round(4).to_string(index=False))

results.round(4).to_csv('data/comparison_metrics.csv', index=False)
```

# 7. Confusion Matrices


```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for ax, y_pred, title in zip(
    axes,
    [y_pred_linear, y_pred_rbf, y_pred_lr],
    ['SVM Linear', 'SVM RBF', 'Regressão Logística']
):
    cm = confusion_matrix(y_test, y_pred)
    ConfusionMatrixDisplay(cm, display_labels=raw_data.target_names).plot(ax=ax, cmap='Blues', colorbar=False)
    ax.set_title(title)

plt.suptitle('Matrizes de Confusão — Conjunto de Teste', fontsize=14)
plt.tight_layout()
plt.savefig('data/confusion_matrices.png', dpi=150, bbox_inches='tight')
plt.show()
```

Fronteira de decisão usando `mean radius` e `mean texture`. São as duas features
que mais separam as classes no dataset, então a visualização fica legível.
Os modelos abaixo são retreinados só com essas duas features.


```python
feature1 = 'mean radius'
feature2 = 'mean texture'

f1_idx = list(raw_data.feature_names).index(feature1)
f2_idx = list(raw_data.feature_names).index(feature2)

X_vis = X_train_scaled[:, [f1_idx, f2_idx]]

svm_linear_vis = SVC(kernel='linear', random_state=42)
svm_linear_vis.fit(X_vis, y_train)

svm_rbf_vis = SVC(kernel='rbf', random_state=42)
svm_rbf_vis.fit(X_vis, y_train)

def plot_decision_boundary(model, X, y, title, ax):
    h = 0.02
    x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
    y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
    xx, yy = np.meshgrid(np.arange(x_min, x_max, h),
                         np.arange(y_min, y_max, h))
    Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = Z.reshape(xx.shape)
    ax.contourf(xx, yy, Z, alpha=0.3, cmap='coolwarm')
    ax.scatter(X[:, 0], X[:, 1], c=y, cmap='coolwarm', edgecolors='k', s=30)
    ax.set_xlabel(feature1)
    ax.set_ylabel(feature2)
    ax.set_title(title)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
plot_decision_boundary(svm_linear_vis, X_vis, y_train, 'SVM Linear — Fronteira de Decisão', axes[0])
plot_decision_boundary(svm_rbf_vis, X_vis, y_train, 'SVM RBF — Fronteira de Decisão', axes[1])
plt.suptitle(f'Fronteira de Decisão: {feature1} vs {feature2}', fontsize=14)
plt.tight_layout()
plt.savefig('data/decision_boundary.png', dpi=150, bbox_inches='tight')
plt.show()
```

# 9. Análise dos Resultados

### 9.1 Desempenho dos três modelos

Fiquei surpreso com o resultado: a regressão logística foi o modelo que teve o melhor desempenho, e não o SVM. A tabela de comparação mostra a regressão logística com 0.9883 de acurácia, o SVM linear com 0.9825 e o SVM RBF com 0.9766. A diferença entre os três é de menos de 1.2%, então nenhum deles se destaca de forma decisiva.

Isso faz sentido quando eu penso na natureza do dataset. São 30 features numéricas bem padronizadas, com uma separabilidade alta entre as classes, e um dataset pequeno (569 amostras). Em um cenário assim, um modelo linear já resolve quase tudo, e o teto de desempenho é limitado pelos casos que realmente se sobrepõem, não pela complexidade do classificador.

### 9.2 Olhando para a classe Malignante

Como o contexto é médico, eu me preocupei menos com a acurácia geral e mais com a classe Malignante (classe 0), que é a minoritária com 64 casos no teste. Um falso negativo aqui significa classificar um tumor maligno como benigno, que é o erro mais grave possível nesse tipo de problema.

O recall maligno ficou em 0.9688 tanto no SVM linear quanto no RBF, e 0.9844 na regressão logística. Só olhando recall, os três modelos estão praticamente empatados, e o único falso negativo em cada um deles é o tipo de caso que nenhuma fronteira de decisão-linear consegue separar sem errar.

A precisão maligno também ficou alta: 0.9841 no linear, 0.9688 no RBF e 0.9844 na logística. Isso quer dizer que os modelos não estão simplesmente classificando tudo como benigno para acertar a acurácia, que era meu principal receio.

### 9.3 Linear vs RBF

O kernel linear procura um hiperplano que maximiza a margem entre as classes, e usa 31 vetores de suporte (14 da classe Malignante e 17 da Benign). Já o RBF mapeia os dados para um espaço de dimensionalidade infinita com `K(x, x') = exp(-γ ||x - x'||²)`, o que permite fronteiras curvas.

O resultado me mostrou algo que eu não esperava: o RBF ficou um pouco **pior** que o linear em acurácia (0.9766 contra 0.9825) e usou quase três vezes mais vetores de suporte (91 contra 31). O motivo é o `gamma` padrão do scikit-learn, que caiu em 0.0333 por causa do escalonamento. Esse valor é baixo demais para o volume do dataset: a influência de cada ponto de treinamento fica espalhada demais, o modelo precisa de muitos vetores de suporte para descrever a fronteira e acaba capturando ruído em vez de estrutura real.

Já o parâmetro `C=1.0` (nos dois modelos) define o trade-off entre margem larga e poucos erros. Com esse valor padrão, o SVM aceita alguns erros de treino para conseguir uma margem maior, o que explica parte da folga nas matrizes de confusão.

### 9.4 Comparação com a Regressão Logística

Vale comparar com o que eu fiz no lab 2. Lá a regressão logística com as 30 features chegou a 0.9737 de acurácia usando `test_size=0.2` sem estratificação. Aqui, usando `test_size=0.3` com `stratify=y`, a mesma técnica chegou a 0.9883.

A melhora não veio do modelo em si, e sim da forma de montar a divisão. Com estratificação, as proporções de 212 Malignant e 357 Benign são mantidas nos dois conjuntos, o que dá uma estimativa mais confiável, especialmente para a classe minoritária. Com apenas 114 amostras de teste no lab 2, um único erro já valia quase 1% de acurácia.

As duas famílias se comportam de forma parecida porque as duas procuram um separador linear. A diferença está no objetivo: a regressão logística otimiza verossimilhança e entrega probabilidades calibradas de graça, enquanto o SVM otimiza margem e entrega uma fronteira mais estável quando existe outliers, já que só os vetores de suporte pesam. Isso fez o SVM perder aqui, porque com 569 amostras limpas não há tanto outlier para o SVM se defender.

### 9.5 Fronteira de Decisão

Na visualização com `mean radius` contra `mean texture`, o SVM linear desenhou uma reta e o RBF desenhou uma curva. Dá para ver que as duas classes se sobrepõem bastante nessa projeção 2D, com casos malignos bem parecidos com benignos. Como o dataset só tem 64 e 107 amostras de teste e nenhuma separação perfeita, os gráficos de fronteira só são aproximações; os números do classification report são a fonte confiável.

### 9.6 Conclusão

Para este dataset, a regressão logística foi a melhor escolha entre as três: mais acurácia, menos falsos negativos e menos vetores de suporte para competir. O SVM linear ficou muito perto e é mais robusto a outliers, o que faria eu escolher ele se o dataset fosse maior ou mais ruidoso. O RBF foi o pior dos três com os parâmetros padrão, e eu entendo o motivo depois de ver que ele usou 91 vetores de suporte com um `gamma` baixo demais.

As limitações que eu encontrei nesse trabalho: os números vêm de um único split fixo, então não sei o quanto variariam em outra divisão, e eu não explorei valores de `C` e `gamma` para o RBF, que provavelmente mudaria o resultado dele. Não consegui gerar probabilidades com o SVM porque o `SVC` não faz isso nativamente, precisaria de `Platt Scaling` ou `cross_val_predict` para calibrar. E a fronteira de decisão foi mostrada em apenas duas features das 30 disponíveis, então ela é apenas ilustrativa e não representa o modelo completo.
